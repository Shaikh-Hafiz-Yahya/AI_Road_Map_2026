## Method overloading and Method overriding

```python
# -----------------------Method Overloaing-----------------------
# Python does not support traditional method overloading i.e Having multiple methods with same name with
# multiple number of arguments
# Although similar functionality can be acheived 
# Using default arguments and variable length arguments

class Calculator: #child/sub class

    def add(self , x , y):

        return x+y
    
    def add(self , x , y , z): #Error (3rd argument fix nahi)

        return x+y+z

c:Calculator = Calculator()
c.add(2 , 7)
```

```python
# Method overloading using method:1 set default argumnent
class Calculator: #child/sub class

    def add(self , x , y):

        return x+y
    
    def add(self , x , y , z=0): #using method1: set default argument

        return x+y+z

c:Calculator = Calculator()
c.add(2 , 7)
```

```python
class Calculator: #child/sub class

    def add(self , *args): #using method2: var length argument
        
        total = 0
        for num in args:

            total = total + num

        return total  

c:Calculator = Calculator()
c.add(2 , 22 , 11 , 8)
```

```python
# --------------------Method Overriding----------------------
# Concept in object oriented programming that allows a subclass (child class) to provide a specific
# implementation for a method that is already defined in it superclass (parent class)

class p:

    def method(self):
        print('I am method inside parent class')

class c(p):
    
    def method(self):
       super().method()
       print('I am method inside child class')

child:c = c()

child.method()
```

```python
class Vehicle:

    def engine(self , model , price , category):

        detail = model+price+category
        print('Vehicles',detail)

class Cars(Vehicle):

    def engine(self, model, price, category):
        super().engine(model , price , category)
    
        c = model+price+category
        print('Text=',c/3)

v:Cars = Cars()
v.engine(2025 , 2000000 , 8)    
```

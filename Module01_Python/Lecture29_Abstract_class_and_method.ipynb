```python
from abc import ABC , abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
    
    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):

    def start(self):
        print('CAR IS STARTED...')
    
    def stop(self):
        print('CAR IS STOPED...')

travel:Car = Car()
travel.start()
travel.stop()
```

```python
from abc import ABC , abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
   
    @abstractmethod 
    def stop(self):
        pass
class Cars(Vehicle):

    def start(self):
        print('CAR IS STARTED...')

    def stop(self):
        print('CAR IS STOPED...')

class Bikes(Vehicle):

    def start(self):
        print('PRESS kICK --> BIKE IS STARTED..')

    def accelerate(self):
        print('NOW VELOCITY MINIMAL INCREASE..')

    def clutch(self):
        print('PRESS..')

    def gear(self):
        print('PRESS FIRST GEAR (Velovity increase = Distance Cover)....')            

    def stop(self):
        print('BIKE IS STOPED (Velocity is 0).')    

travel:Bikes = Bikes()

travel.start()  #abstract method implement kiyay child class main
travel.accelerate() #class k apny method bhi implement kiyay
travel.clutch() #class k apny method bhi implement kiyay
travel.gear() #class k apny method bhi implement kiyay
travel.stop() #abstract method implement kiyay child class main
```

```python
from abc import ABC , abstractmethod
class Machine(ABC):

    @abstractmethod
    def thinking(self):
        pass

    @abstractmethod
    def action(self):
        pass

class Agents(Machine):

    def thinking(self):
        print('Decision')

    def decision(self):  #child ka apna method implement kiya    
        print('Satisfaction') 

    def action(self):
        print('Complete Any Task (Action Perform LIKE Human)') 

ai:Agents = Agents()

ai.thinking()
ai.decision()
ai.action()
```

Duck Typing

```python
class Birds:

    SkyBeauti = 'All Birds Can Fly In The Sky'

class Parrots(Birds):

    def makesound(self):
        print('Turrr Turrr')

class Piegon(Birds):

    def makesound(self):
        print('Gurr Gurr')

class Sparrow(Birds):

    def makesound(self):
        print('Pee Pee')  
 
    @classmethod
    def makesound(cls):
        print(cls.SkyBeauti)  

Sky = [Parrots() , Piegon() , Sparrow()]
for fly in Sky:
    fly.makesound()      
                                
```

```python
class Doctor:

    treatment = True

class Patient1(Doctor):

    def unfit(self):
        print('Patient --> high fever')

class Patient2(Doctor):

    def unfit(self):
        print('Patient --> heart disease')      

class HealthyPerson:

    def fit(self):
        print('Health --> 90%') 

class FitAndHealthy:
    
    def fit(self):
        print('FitHealthy --> 98.99%')

treat = [Patient1() , Patient2()] #pure Duck Typing nhi ha(ye Inheritance + Polymorphism)but Duck Typing ha

f = [HealthyPerson() , FitAndHealthy()] #pure Duck Typing 100% (Raeson:Absence of inheritance)

for p in treat:
    p.unfit()

for enjoy_life in f:
    enjoy_life.fit()    
```


# Operator Overloading In Polymorphism

operator overloading in python allow yo to redefine how built-in operators(+,-,*,etc) behave for user defined objects(classes) .This can be acheived by special by sppecial functions like __add__ , __sub__ stc

```python
x , y = 20 , 12
print(x+y)
```

```python
x , y = '20' , '12'
print(x+y) #Concatenation (ye hi operator overloading ka concept ha)

# similarly

print(str.__add__(x , y))
```

```python
x , y = 20 , 12
print(x+y) 

# similarly

print(int.__add__(x , y))
```

```python
class Invoice:

    def __init__(self , invoiceamount):

        self.amount = invoiceamount

    #Overloading
    def __add__(self, other):

        total = self.amount + other.amount
        print(total)

invo1:Invoice = Invoice(2000)
invo2:Invoice = Invoice(10000)   
invo1 + invo2   
```

```python
class MyInvoice:

    def __init__(self , invoiceamount):

        self.amount = invoiceamount
    
    #Overloading
    def __sub__(self, other):
        
        difference = self.amount - other.amount
        print(difference)

inv1:MyInvoice = MyInvoice(12000)
inv2:MyInvoice = MyInvoice(2000)

inv1-inv2        
```

```python
class Product:

    def __init__(self , num):
        self.intnum = num

    def __mul__(self , other):
        mult = self.intnum * other.num 
        print(mult)

num1:Product = Product(20)
num2:Product = Product(40)

num1.intnum*num2.intnum
```

```python
class Decimal:

    def __init__(self , num):
        self.decimal = num

    def __add__(self, other):
        summation = self.decimal + other.num
        print(summation)

d1:Decimal = Decimal(2.11)
d2:Decimal = Decimal(3.1415)

d1.decimal + d2.decimal
         
```

```python
class Subtract:

    def __init__(self , num):
        self.num = num

    def __sub__(self, other):
        Subtract_num = self.num + other.num
        print(Subtract_num)

s1:Subtract = Subtract(12)
s2:Subtract = Subtract(15)


s1.num - s2.num          
```

```python
class Linear:

    def __init__(self , m , b):

        self.slop = m
        self.bias = b

    def __add__(self, other):

        y = self.slop + other.bias
        print(y)

l1:Linear = Linear(500 , 1000)
l2:Linear = Linear(500 , 10000)

four_days_work = l1.slop + l2.bias
six_days_work = l1.slop + l2.bias
thirty_days_work = l2.slop + l2.bias
three_sixty_days_work = l2.slop + l2.bias

print(f'Employee Salary {four_days_work}RS Only..')
print(f'Employee Salary {six_days_work}RS Only..')
print(f'Employee Salary {thirty_days_work}RS Only..')
print(f'Employee Salary {three_sixty_days_work}RS Only..')
```

```python
class Company:

    def __init__(self, m, b):

        self.slope = m
        self.bias = b

    def __add__(self, other):

        return Company(
            self.slope + other.slope,
            self.bias + other.bias
        )
    
c1:Company =Company(500 , 1000)
c2:Company = Company(5000 , 10000)    

x=[30,100,200,360]

for days in x:
    salary = c2.bias+days*c2.slope
    e = {days:salary} 
    
    if salary > 1500000:
        print('CEO')
    print(f'Employee:{e}=WorkingDays : Salary\n')
print(type(e))
```

# **List Comprehension**

````python
# without List Comprehension
lst = [20 , 21 , 22 , 23 , 24]

emp_lst = []

for i in lst:

    emp_lst.append(i*0.25)

emp_lst    
````

````python
# Now Using List Comprehension
[i*0.25 for i in lst]
````

````python
[a*2 for a in range(100-90)]
````

````python
fruit_box = [apple*3 for apple in range(1 , 10)]
fruit_box
````

````python
even_num = [even for even in range(1 , 31) if even%2==0]
even_num
````

````python
odd_num = [odd for odd in range(30) if odd%2 == 1]
odd_num
````

Generating Data From Data Using List Comprehension

````python
df7 = pd.DataFrame(

    {
        'Apple':[100 , 2010 , 1900 , 1001],
        'Banana':[200 , 1999 , 100 , 2006],
        'Orange':[100 , 199 , 1010 , 206]

    },

    index=['Tree01', 'Tree02', 'Tree03' , 'Tree04']

)

df7
````

````python
df7['Guava'] = [n+n*0.95 if n<=100 else n>100 if n==1001 else n>1001 for n in df7['Apple']]
df7
````

````python
df7['Guava'] = [n+n*0.95 if n>=1000 else n-n*0.5 if n<1000 else n-n*0.9 for n in df7['Apple']]
df7
````

````python
df7['Cherry'] = [n+n*0.97 
                 if n>=50.99 
                 else n+n*1.99 
                 for n in df7['Guava']]
df7
````

````python
df7['Total'] = df7['Apple'] + df7['Banana'] + df7['Orange'] + df7['Guava'] + df7['Cherry'] / 6
df7
````

````python
df7['percentage'] = round(df7['Total']/10000*100 , 2)
df7
````

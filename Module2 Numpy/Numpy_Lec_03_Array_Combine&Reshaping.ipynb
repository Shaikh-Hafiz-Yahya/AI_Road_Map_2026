# **NumPy Array Combine & Reshaping**

```python
import numpy as np
```

```python
array2d = np.arange(64).reshape(8 , 8)
array2d
```

```python
array2d > 50
```

```python
array2d[array2d > 50]
```

```python
array2d[array2d%5 == 0]
```

```python
array2d[array2d%7 == 1]
```

```python
array2d[array2d%2 > 0] #Odd Numbers
```

```python
array2d[array2d%2 == 0] #Even Numbers
```

```python
array2d[array2d%2 > 0] + array2d[array2d%2 == 0]
```

```python
array2d[(array2d%5 == 0) & (array2d%7 != 0)]
```

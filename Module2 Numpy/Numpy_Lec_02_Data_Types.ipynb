# Numpy Lecture 02 - Data Types

```python
import numpy as np
```

```python
arr = np.array([2 , 4 , 6 , 8] , dtype=np.int8)

print(arr)
print(arr.dtype)
print(arr.nbytes)
```

```python
arr2 = np.array([2 , 4 , 6 , 8] , dtype=np.float32)

print(arr2)
print(arr2.dtype)
print(arr2.nbytes)
```

```python
arr3 = np.array([2 , 3 , 4 , 5] , dtype=np.bool_)

print(arr3)
print(arr3.dtype)
print(arr3.nbytes)
```

```python
arr4 = np.array([1 , 3 , 5 , 7] , dtype=np.str_)

print(arr4)
print(f'DataType are single digit = {arr4.dtype}')
print(arr4.nbytes)
```

# Diff B/W

```python
arr5 = np.array([11 , 33, 56 , 79] , dtype=np.str_)

print(arr5)
print(f'DataType are double digits = {arr5.dtype}')
print(arr5.nbytes)
```

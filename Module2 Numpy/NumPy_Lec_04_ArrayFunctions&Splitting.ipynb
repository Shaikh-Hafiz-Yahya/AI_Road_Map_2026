### **1. Splitting Arrays**

```python
import numpy as np
```

```python
a = np.arange(12).reshape(3 , 4)
a
```

```python
axis0 = np.vsplit(a , 3)
axis0
```

```python
axis1 = np.hsplit(a , 4)
axis1
```

```python
axis1[0]
```

```python
axis1[0].ndim
```

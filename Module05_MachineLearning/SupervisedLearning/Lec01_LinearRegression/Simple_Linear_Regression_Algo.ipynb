# *`SIMPLE LINEAR REGRESSION ALGORITHM`*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv(r"C:\Users\Muhammad Yahya\Downloads\python\AI Roadmap2026\AI_Road_Map_2026\Module05_MachineLearning\SupervisedLearning\Lec01_LinearRegression\test.csv")
df.head()
```

# **Visualization**

```python
plt.scatter(df.x , df.y)
plt.grid()
plt.xlabel("Input Feature(x=indep)" , color='r')
plt.ylabel("Output Feature(y=dep/Label/Target)" , color='r')
```

# **Build Linear Regression Basic/Small Model**

```python
x = df[['x']]
y = df['y']
lin_reg_model = LinearRegression()
lin_reg_model = lin_reg_model.fit(x , y) 
lin_reg_model
```

# **Prediction**

```python
predict = lin_reg_model.predict([[21.1]])
print(f"Model Prediction(Target) = {predict}")
```

```python
lin_reg_model.score(x , y)
```

# **Coefficient/Slop**

```python
coeffi =  lin_reg_model.coef_
coeffi
```

# **Intercept**

```python
intercept = lin_reg_model.intercept_
intercept
```

# **Simple Linear Regression Formula!**

```python
m = coeffi
b = intercept
x = 21.1

y = m*x + b 
y
```

```python
prediction = y
prediction
```

# **`Verify Manual Calculation`**

```python
if (y == predict):
    
    print(f"Manual Calculation is {True}\ny={y} = predict={predict}")
    plt.scatter(x , y , color='r')
    plt.grid()

else:
    print(f"Manual Calculation is Not Verify {False}")    
```
